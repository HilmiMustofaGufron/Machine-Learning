```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/febbisena/DataMining/refs/heads/main/milkgrading.csv')
df
```

```python
en = LabelEncoder()

df['Grade'] = en.fit_transform(df['Grade'])
df.head()
```

```python
label_maping = pd.Series(en.classes_)
label_maping
```

```python
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
x
```

```python
y
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, random_state=50)

print("x_train = ", len(x_train))
print("x_test = ", len(x_test))
print("y_train = ", len(y_train))
print("y_test = ", len(y_test))
```

```python
test_df = pd.DataFrame(x_test, columns=df.columns[:-1])
test_df['Actual'] = y_test
test_df
```

```python
sc = StandardScaler()

x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)

x_train
```

```python
nb = GaussianNB()
nb.fit(x_train, y_train)
```

```python
y_pred = nb.predict(x_test)
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
plt.figure(figsize=(5,4))
sns.set(font_scale=1.4)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=en.classes_, yticklabels=en.classes_)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')
plt.show()
```

```python
creport = classification_report(y_test, y_pred)
print(creport)
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print("Tingkat Akurasi : %.2f persen" % (accuracy * 100))
```

```python
actual_counts = pd.Series(y_test).value_counts().sort_index()
predicted_counts = pd.Series(y_pred).value_counts().sort_index()
```

```python
actual_counts
```

```python
predicted_counts
```

```python
plt.figure(figsize=(6, 6))
sns.set(font_scale=1)
width = 0.4
classes = range(len(en.classes_))

bars1 = plt.bar(classes, actual_counts, width, label='Actual')
bars2 = plt.bar([p + width for p in classes], predicted_counts, width, label='Predicted')

plt.bar_label(bars1, fmt='%d')
plt.bar_label(bars2, fmt='%d')

plt.legend()

plt.xlabel('Class')
plt.ylabel('Count')
plt.title('Actual vs. Predicted Counts')
plt.xticks([p + width/2 for p in classes], en.inverse_transform(classes))

plt.show()
```
